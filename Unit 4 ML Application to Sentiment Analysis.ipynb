```python
class SimpleRNNNeuron:
    def __init__(self, input_size, hidden_size, learning_rate=0.1):
        self.weights_input_hidden = np.random.rand(input_size, hidden_size)
        self.weights_hidden_hidden = np.random.rand(hidden_size, hidden_size) # Recurrent connection
        self.bias_hidden = np.random.rand(hidden_size)
        self.learning_rate = learning_rate
        self.hidden_state = np.zeros(hidden_size) # Initial hidden state

    def tanh(self, x):
        return np.tanh(x)

    def process_sequence(self, sequence):
        outputs = []
        for input_t in sequence:
            hidden_layer_input = np.dot(input_t, self.weights_input_hidden) + \
                                 np.dot(self.hidden_state, self.weights_hidden_hidden) + \
                                 self.bias_hidden
            self.hidden_state = self.tanh(hidden_layer_input)
            outputs.append(self.hidden_state)
        return outputs
```
